# Student Perfoemance Dataset Cleaning and Manipulation

```python
# Import necessary libraries
import pandas as pd

# Load the dataset
data = pd.read_csv("StudentsPerformance.csv")

# Display the first few rows to understand the data
data.head()
```

### Handle Missing Values

```python
# Print sum of missing values
data.isnull().sum()

#no missing values found
```

### Handle Duplicate Values

```python
#Check for the duplicates
data.duplicated().sum()

#no duplicates found
```

### Filltering The Dataset

```python
# Filter to include only students with 'math score' above the median
median_math_score = data['math score'].median()
filtered_df = data[data['math score'] > median_math_score]
print(filtered_df)
```

```python
# Count gender occurrences within each race/ethnicity group
data.groupby('race/ethnicity')['gender'].value_counts()
```

```python
# Count occurrences of each parental education level within each race/ethnicity group
data.groupby('race/ethnicity')['parental level of education'].value_counts()
```

```python
# Count students with bad math scores below 50, grouped by test preparation course, sorted in descending order
bad_math =data[data['math score']<50]
bad_math.groupby('test preparation course')['math score'].count().sort_values(ascending=False)
```

### Dataset Feature Engineering

```python
# Add a new column 'average score' as the mean of the three score columns
data['average score'] = data[['math score', 'reading score', 'writing score']].mean(axis=1)
print(data.head())
```
